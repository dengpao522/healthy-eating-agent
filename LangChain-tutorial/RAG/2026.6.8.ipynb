```python

from langchain_community.document_loaders import TextLoader
#指明txt文档的路径（文件就在当前RAG目录下）
file_path = './langchain-utf-8.txt'

#创建一个TextLoader的实例
text_loader = TextLoader(
    file_path=file_path,
    encoding='utf-8',
)

#调用load() 返回一个list
docs = text_loader.load()

# #打印文档数量
# print(f"加载了 {len(docs)} 个文档")
#
# #查看第一个文档的内容（前500个字符）
# if len(docs) > 0:
#     print("\n文档内容预览：")
#     print(docs[0].page_content[:500])

print(docs)

#显示Document的元数据
print(docs[0].metadata)

#显示文档中的内容信息
print(docs[0].page_content)
```
