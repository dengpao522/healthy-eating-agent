```python
from langchain_classic.memory import ConversationBufferMemory
##%%
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor
from langchain_core.prompts import ChatPromptTemplate
# 获取Tavily搜索的实例
from langchain_openai import ChatOpenAI

import os
import dotenv
from langchain_openai import ChatOpenAI
from langchain_community.tools.tavily_search import TavilySearchResults
dotenv.load_dotenv()

# 读取配置文件的信息
#Tavily API Key 已从 ..env 文件中自动加载，无需手动设置
# os.environ['TAVILY_API_KEY'] = "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"

# 获取Tavily搜索工具的实例（TavilySearchResults 本身就是工具，无需额外包装）
search_tool = TavilySearchResults(max_results=3)

os.environ['OPENAI_API_KEY'] = os.getenv("OPENAI_API_KEY")
os.environ['OPENAI_BASE_URL'] = os.getenv("OPENAI_BASE_URL")

llm = ChatOpenAI(model='qwen-turbo',
                 temperature=0
                 )


# 提供提示词模板（以ChatPromptTemplate为例）
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个乐于助人的ai助手，根据用户的提问，必要时调用Search工具，使用互联网检索数据"),

    ("system", "{agent_scratchpad}"),
    ("system", "{chat_history}"),
    ("human", "{input}"),
])

# 提供记忆的实例：ConversationBufferMemory
memory = ConversationBufferMemory(
    return_messages=True,
    memory_key="chat_history", #此值需要与提示模板中的记录记忆的变量同名
)


# 获取Agent的实例：create_tool_calling_agent()
agent = create_tool_calling_agent(
    llm=llm,
    prompt=prompt_template,
    tools=[search_tool]
)

# 获取AgentExecutor的实例
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose= True,
    memory = memory
)

# 通过AgentExecutor的实例调用invoke()，得到响应
result = agent_executor.invoke({"input":"查询今天北京的天气情况"})
result1 = agent_executor.invoke({"input":"上海的呢"})
# 处理响应
print(result)
print(result1)

```
